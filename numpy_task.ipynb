```python
import numpy as np
```

```python
#task 1
marks= np.array([10, 20, 30, 40, 50])
print("\nAverage Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print(np.sum(marks))
print(marks)
```

```python
#task 2
marks= np.array([45, 80, 30, 90, 55, 20])
print(marks[marks >= 50])
```

```python
#task 3
marks = np.array([1, 2, 3, 4, 5, 6])
new_array = marks.reshape(2, 3)

print(new_array)
```

```python
#task 4
marks = np.random.randint(40, 101, 10)

print("Random Marks:", marks)
print("Average:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
#task 5
marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])
passed = marks[marks >= 50]
failed = marks[marks < 50]

print("Passed Marks:", passed)
print("Failed Marks:", failed)

print("Number of Passed Students:", len(passed))
print("Number of Failed Students:", len(failed))
```

```python
#practice
#ndim
arr = np.array([10, 20, 30, 40, 50])

print("Dimensions:", arr.ndim)
```

```python
#shape
marks = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Shape:", marks.shape)
```

```python
#size
marks = np.array([10, 20, 30, 40, 50])

print("Total Elements:", marks.size)
```

```python
#dtype
marks = np.array([10, 20, 30, 40, 50])

print("Data Type:", marks.dtype)
```

```python
#indexing
marks = np.array([50, 60, 70, 80, 90])

print("First Mark:", marks[0])
print("Third Mark:", marks[2])
print("Last Mark:", marks[-1])
```

```python
#slicing
marks = np.array([10, 20, 30, 40, 50, 60])

print("Sliced Array:", marks[1:4])
```

```python
#vectorized operation
marks = np.array([10, 20, 30, 40, 50])

print("Add 5:", marks + 5)
print("Multiply by 2:", marks * 2)
print("Subtract 5:", marks - 5)
```

```python
#flatten
marks = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("2D Array:")
print(marks)

print("Flatten Array:")
print(marks.flatten())
```
